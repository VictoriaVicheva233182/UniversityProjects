```python
import requests
import pandas as pd
import concurrent.futures
from sklearn.metrics import classification_report, accuracy_score, f1_score

API_BASE = "http://194.171.191.228:30434"

# Minimized prompt with diverse examples
PROMPT_TEMPLATE = """
Classify the emotion as one of: happiness, sadness, anger, surprise, fear, disgust, neutral.

Examples:
"I am so excited!" → happiness
"I don't know where I am." → surprise
"I feel so sad right now." → sadness
"There are dangerous snakes here." → fear
"This is unfair, I'm angry!" → anger
"I feel sick just thinking about it." → disgust
"The weather is normal today." → neutral

Sentence: "{sentence}"
Answer:
"""

def get_emotion(sentence: str) -> str:
    """Get emotion classification with a minimal prompt."""
    prompt = PROMPT_TEMPLATE.format(sentence=sentence)
    
    try:
        response = requests.post(
            f"{API_BASE}/api/chat",
            json={
                "model": "llama3.3",
                "messages": [{"role": "user", "content": prompt}],
                "stream": False,
            },
            timeout=1000  # Reduce timeout for faster processing
        )
        result = response.json()
        emotion = result.get('message', {}).get('content', '').strip().lower()

        # Validate and clean predictions
        valid_emotions = {'happiness', 'sadness', 'anger', 'surprise', 'fear', 'disgust', 'neutral'}
        return emotion if emotion in valid_emotions else 'neutral'
    
    except Exception as e:
        print(f"Error processing sentence: {str(e)}")
        return "neutral"

def worker(sentence):
    """Wrapper function for parallel processing."""
    return get_emotion(sentence)

def process_dataset(csv_path: str, num_rows: int = 1000):
    """Process dataset using parallel API calls and calculate metrics."""
    try:
        df = pd.read_csv(csv_path).head(num_rows).dropna(subset=['Translation', 'Emotion'])
        true_labels = df['Emotion'].str.lower().tolist()
        
        # Parallel processing for API calls
        with concurrent.futures.ThreadPoolExecutor(max_workers=2000) as executor:
            pred_labels = list(executor.map(worker, df['Translation']))
        
        print(f"Accuracy: {accuracy_score(true_labels, pred_labels):.2f}")
        print(f"F1 Score: {f1_score(true_labels, pred_labels, average='weighted'):.2f}")
        print("\nClassification Report:")
        print(classification_report(true_labels, pred_labels))
        
    except Exception as e:
        print(f"Dataset error: {str(e)}")

def main():
    csv_path = r"C:\Users\Edopi\Desktop\2024-25c-fai2-adsai-EdoardoPierezza231412\Datalab_task\Task5\Processed\processed_group 9_url1.csv"
    process_dataset(csv_path, num_rows=1000)

if __name__ == "__main__":
    main()
```
