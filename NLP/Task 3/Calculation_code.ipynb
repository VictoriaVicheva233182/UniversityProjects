```python
import pandas as pd
import re


file_path_whisper = r"C:\Users\Edopi\Desktop\2024-25c-fai2-adsai-group-team_9\Data_lab_task\Task2\transcribed_data_assemblyAI.csv"
df_whisper = pd.read_csv(file_path_whisper)

# Given substitution, deletion, and insertion values
S_whisper = 1   # Substitutions
D_whisper = 49  # Deletions
I_whisper = 72  # Insertions

# Function to tokenize text using regex (splitting on words, punctuation, and numbers)
def tokenize_text(text):
    return re.findall(r"\b\w+\b|[^\w\s]", str(text))  # Tokenizing words, punctuation, and numbers

# Apply tokenization to each transcription
df_whisper["Tokenized"] = df_whisper["Sentence"].apply(tokenize_text)

# Count the total number of tokens in the transcribed text
token_count_in_whisper = df_whisper["Tokenized"].str.len().sum()

# Estimate the correct N (ground truth token count)
N_whisper_corrected = token_count_in_whisper + D_whisper - I_whisper

# Calculate WER using tokens
WER_whisper_corrected = (S_whisper + D_whisper + I_whisper) / N_whisper_corrected

# Print the results
print(f"Total token count in AssemblyAI transcription: {token_count_in_whisper}")
print(f"Estimated N (Ground Truth Token Count): {N_whisper_corrected}")
print(f"Word Error Rate (WER): {WER_whisper_corrected:.4f}")
```
