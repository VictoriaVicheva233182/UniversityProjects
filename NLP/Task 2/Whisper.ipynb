```python
# importing packages
from pytubefix import YouTube
import os
```

```python

# url input from youtube
yt = YouTube("https://www.youtube.com/watch?v=zriFfGkQ2dY")

# extract only audio
video = yt.streams.filter(only_audio=True).first()

# set destination to save file
destination = ("C:/Users/kosta/OneDrive/Documents/GitHub/2024-25c-fai2-adsai-KostaLjubisavljevic233101/Datalab tasks/Task2")

# download the file
out_file = video.download(output_path=destination)

# save the file
base, ext = os.path.splitext(out_file)
new_file = base + '.mp3'
os.rename(out_file, new_file)
```

```python
import whisper
import pandas as pd

# Load the Whisper model
model = whisper.load_model("turbo")

# Transcribe the entire audio file
result = model.transcribe("Survivor NZ  Season 1 (2016)  Episode 1 - FULL EPISODE.mp3")

# Print the full transcription
print(result["text"])

# Split the text into lines (based on punctuation for better readability)
lines = result["text"].split(". ")  # Splitting at sentence boundaries

# Save each line in a CSV file
df = pd.DataFrame(lines, columns=["Transcription"])
df.to_csv("transcriptturbo.csv", index=False)
```
