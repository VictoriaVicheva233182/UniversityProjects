```python
from transformers import MarianMTModel, MarianTokenizer
import pandas as pd
import torch

def load_model(lang_pair):
    """Load MarianMT model and tokenizer for translation"""
    model_name = f"Helsinki-NLP/opus-mt-{lang_pair}"
    tokenizer = MarianTokenizer.from_pretrained(model_name)
    model = MarianMTModel.from_pretrained(model_name)

    # Move model to GPU if available
    device = "cuda" if torch.cuda.is_available() else "cpu"
    model.to(device)

    return model, tokenizer, device

def translate_text_in_batches(texts, model, tokenizer, device, batch_size=64):
    """Translate text in batches to optimize performance"""
    translated_texts = []
    
    for i in range(0, len(texts), batch_size):
        batch = texts[i:i + batch_size]
        inputs = tokenizer(batch, return_tensors="pt", padding=True, truncation=True).to(device)
        translated = model.generate(**inputs)
        translated_batch = [tokenizer.decode(t, skip_special_tokens=True) for t in translated]
        translated_texts.extend(translated_batch)

        print(f"Processed {i + len(batch)}/{len(texts)} sentences...")

    return translated_texts

# Load models
model_en_it, tokenizer_en_it, device = load_model("en-it")  # English → Italian
model_it_en, tokenizer_it_en, device = load_model("it-en")  # Italian → English

# Load transcript file
file_path = "transcribed_data_assemblyAI.csv"
df = pd.read_csv(file_path)

# Ensure the correct column name exists
if "Sentence" not in df.columns:
    raise ValueError("Column 'Sentence' not found in the CSV file. Check column names.")

english_sentences = df["Sentence"].tolist()

# Translate English → Italian in batches
print("Translating English to Italian...")
italian_sentences = translate_text_in_batches(english_sentences, model_en_it, tokenizer_en_it, device, batch_size=64)

# Translate Italian → English (Back Translation) in batches
print("Translating Italian back to English...")
back_translated_sentences = translate_text_in_batches(italian_sentences, model_it_en, tokenizer_it_en, device, batch_size=64)

# Save results to DataFrame
df["Translation"] = italian_sentences
df["English Translation"] = back_translated_sentences

# Save to CSV
output_file = "translated_transcript.csv"
df.to_csv(output_file, index=False)

print(f"Translation complete! Check '{output_file}'")


```
