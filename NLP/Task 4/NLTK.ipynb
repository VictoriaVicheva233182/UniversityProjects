```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from textblob import TextBlob
from nltk.tokenize import sent_tokenize
from gensim.models import Word2Vec
import numpy as np
from nltk.sentiment import SentimentIntensityAnalyzer
```

```python
import nltk
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
# Load transcript CSV
df = pd.read_csv("C:/Users/kosta/OneDrive/Documents/GitHub/2024-25c-fai2-adsai-group-team_9/Data_lab_task/Task2/transcribed_data_assemblyAI.csv")

# Part-of-Speech (POS) Tagging
def get_pos_tags(text):
    return [(word, TextBlob(word).tags[0][1]) for word in text.split()]

df['POS_Tags'] = df['Sentence'].apply(get_pos_tags)
```

```python
# TF-IDF Calculation
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(df['Sentence'])
df['TF_IDF'] = list(tfidf_matrix.toarray())

# Sentiment Analysis
sia = SentimentIntensityAnalyzer()
def get_sentiment(text):
    return sia.polarity_scores(text)['compound']

df['Sentiment_Score'] = df['Sentence'].apply(get_sentiment)
```

```python
# Custom Word Embedding Model
sentences_tokenized = [sent_tokenize(sentence) for sentence in df['Sentence']]
tokenized_words = [[word for sentence in sent_group for word in sentence.split()] for sent_group in sentences_tokenized]
model = Word2Vec(tokenized_words, vector_size=100, window=5, min_count=1, workers=4)
```

```python

def get_custom_embedding(text):
    words = text.split()
    vectors = [model.wv[word] for word in words if word in model.wv]
    return np.mean(vectors, axis=0) if vectors else np.zeros(100)

df['Custom_Embeddings'] = df['Sentence'].apply(get_custom_embedding)

# Additional Feature: Sentence Length
df['Sentence_Length'] = df['Sentence'].apply(lambda x: len(x.split()))

# Save to CSV for Excel
df.to_csv('NLTK_transcript.csv', index=False)

print("Processed CSV saved as 'NLTK_transcript.csv'")
```
