```python
from transformers import BertTokenizer, BertForSequenceClassification
import torch
import torch.nn.functional as F

# Load model and tokenizer from directory paths (directories, not files!)
tokenizer = BertTokenizer.from_pretrained(
    r"D:\YearTwoAI\Block C\2024-25c-fai2-adsai-VictoriaVicheva233182\Week 4\emotion_model_BERT5_tokenizer"
)
model = BertForSequenceClassification.from_pretrained(
    r"D:\YearTwoAI\Block C\2024-25c-fai2-adsai-VictoriaVicheva233182\Week 4\emotion_model_BERT5"
)

# Input sentence
sentence = "Oh, today is such a nice day!"

# Tokenize input
inputs = tokenizer(
    sentence, return_tensors="pt", padding=True, truncation=True, max_length=256
)

# Run the model
model.eval()
with torch.no_grad():
    outputs = model(**inputs)

# Get logits and predicted emotion
logits = outputs.logits
probs = F.softmax(logits, dim=1)
predicted_class = torch.argmax(probs, dim=1).item()

# Map index to label (this must match the label order used during training)
emotion_labels = [
    "happiness",
    "neutral",
    "sadness",
    "anger",
    "surprise",
    "fear",
    "disgust",
]
predicted_emotion = emotion_labels[predicted_class]

print(f"Predicted Emotion: {predicted_emotion}")
```

```python
test_sentences = [
    "Delicious with a drink.",
]

for sentence in test_sentences:
    inputs = tokenizer(
        sentence, return_tensors="pt", padding=True, truncation=True, max_length=256
    )
    with torch.no_grad():
        outputs = model(**inputs)
    logits = outputs.logits
    probs = F.softmax(logits, dim=1)
    predicted_class = torch.argmax(probs, dim=1).item()
    predicted_emotion = emotion_labels[predicted_class]
    print(f"Sentence: {sentence}\nPredicted Emotion: {predicted_emotion}\n")
```
